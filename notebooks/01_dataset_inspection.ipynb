# FOODWISE AI - Dataset Inspection and Annotation Validation
## PROJ3 - Food Waste Detection Dataset (Version 6, Roboflow Universe)
This notebook audits the dataset used for training and evaluating YOLOv11 instance segmentation.

```python
from pathlib import Path
import yaml
import pandas as pd
import matplotlib.pyplot as plt
import cv2
from PIL import Image

DATASET_YAML = Path('../data/processed/dataset.yaml')
with open(DATASET_YAML, 'r') as f:
    cfg = yaml.safe_load(f)

print(f"Total classes: {len(cfg['names'])}")
print("Sample classes:", list(cfg['names'].items())[:10])
```

### 1. Class Distribution Analysis
Inspect the top food classes in the dataset.

```python
df_classes = pd.read_csv('../reports/class_distribution.csv')
print("Top 15 classes by annotation count:")
display(df_classes.head(15))

# Check for rice
rice_rows = df_classes[df_classes['class_name'].str.contains('rice', case=False)]
print("Rice classes verification:")
display(rice_rows)
```

### 2. Annotation Visualizations
Display sample overlays generated during the dataset audit.

```python
vis_dir = Path('../reports/annotation_visualizations')
vis_files = sorted(list(vis_dir.glob('*.jpg')))[:4]

fig, axes = plt.subplots(2, 2, figsize=(14, 14))
for i, ax in enumerate(axes.flatten()):
    if i < len(vis_files):
        img = Image.open(vis_files[i])
        ax.imshow(img)
        ax.set_title(vis_files[i].name, fontsize=10)
        ax.axis('off')
plt.tight_layout()
plt.show()
```
